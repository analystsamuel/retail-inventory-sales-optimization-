```python
df["Weather Condition"] = df["Weather Condition"].fillna("Unknown")
df["Discount"] = df["Discount"].fillna("0")
df["Units Sold"] = df["Units Sold"].fillna("0")
df["Units Ordered"] = df["Units Ordered"].fillna("0")
df["Inventory Level"] = df["Inventory Level"].fillna(df["Inventory Level"].mean())
df["Demand Forecast"] = df["Demand Forecast"].fillna(df["Demand Forecast"].mean())
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df["Category"] = df["Category"].str.strip().str.title()
df["Region"] = df["Region"].str.strip().str.title()
df["Seasonality"] = df["Seasonality"].str.strip().str.title()
```

```python
df["Date"] = pd.to_datetime(df["Date"], format = "mixed", errors = "coerce")
```

```python
df.to_csv("retail_inventory_cleaned.csv", index = False)
```
